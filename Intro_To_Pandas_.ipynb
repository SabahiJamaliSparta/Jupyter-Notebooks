```python
#NaN represents a missing numeric value; it is not the same as None
new_sales["Maya"] is None
```

```python
#Check the entire Series for missing values using pd.isnull()
pd.isnull(new_sales)
```

## DataFrames

DataFrames are 2D, size-mutable, potentially heterogeneous (diverse) tabular data structures. This data structure contains labeled axes (rows and columns).

```python
#Creating a DataFrame from a list of lists
df_data = [["Adrian", 20], ["Max", 23], ["Jax", 41]]

#Provide the column names in the same order as the data
df = pd.DataFrame(df_data, columns=["Name", "Age"])
```

```python
df
```

```python
# Creating a DataFrame from a dictionary 
new_dict2 = {'Name':['Tom','Jane','Steve','Lucy'],'Sales':[250,300,350,400]}

df_dict = pd.DataFrame(new_dict2)
df_dict
```

```python
##Adding a custom index
df_dict_index = pd.DataFrame(new_dict2, index = ["Rank 1", "Rank 2", "Rank 3", "Rank 4"])

df_dict_index
```

```python
# Creating a DataFrame from a list of Dictionaries # This is the same data as we had previously, but in a different format
# This is useful when working with JSON
dict_list = [{'Name':'Tom','Sales':250},{'Name':'Jane','Sales':300},{'Name':'Steve','Sales':350},{'Name':'Lucy','Sales':400}]

df_dict_list = pd.DataFrame(dict_list)
df_dict_list
```

```python
#Creating a DataFrame from a dictionary of series 

east = pd.Series ([1000, 1200, 3400], index =["Q1", "Q2", "Q3"])

east
```

```python
west = pd.Series ([1100, 1300, 2400,3500], index =["Q1", "Q2", "Q3", "Q4"])

west
```

```python
## We can combine our series to create a brand new DataFrame containing both 

df_region = pd.DataFrame({"East":east, "West":west})

df_region
```

```python
#We can add new series into our DataFrame after we have created it!

df_region ["North"] = [2000, 3000, 2500, 4000]

df_region ["South"] = [1500, 2000, 1500, 4000]

df_region
```

```python
#Add a Years column to the DataFrame
years = ["2016", "2017", "2018", "2019"]

df_region["Years"] = years

df_region
```

```python
#Set the Years column as the index, replacing the quarters (Q1 etc.)
df_region = df_region.set_index("Years")

df_region
```

```python
#Reindex the rows and add two new years with NaN values
re_indexed = df_region.reindex(["2016", "2017", "2018", "2019", "2020", "2021"])

#Keep new_df available for the next examples
new_df = re_indexed.copy()

re_indexed
```

## Handling Missing Data

```python
#Filling in missing values
#Some algorithms cannot handle NaN values
#Display missing values replaced with 0 without changing re_indexed
re_indexed.fillna(0)
```

```python
#Forward fill: carry the previous available value down each column
#Backward fill uses the next available value instead: re_indexed.bfill()
#This displays the result without changing re_indexed
re_indexed.ffill()
```

```python
#Linear interpolation estimates missing values between known values
#By default, trailing gaps are filled with the last available value
re_indexed.interpolate()
```
