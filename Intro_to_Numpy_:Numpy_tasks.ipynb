The array contains 10 student scores. `.max()` finds the highest score, `.min()` finds the lowest, `.mean()` calculates the average and `.sum()` calculates the total.

```python
#Highest score
scores.max()
```

```python
#Lowest score
scores.min()
```

```python
#Average score
scores.mean()
```

```python
#Total score
scores.sum()
```

### Bonus Marks

Adding 5 to the array gives each student 5 bonus marks. The result is saved as `bonus_scores`.

```python
#Add 5 bonus marks to each student
bonus_scores = scores + 5

bonus_scores
```

## 2. Weather Data

The array contains daily temperatures in degrees Celsius. Each value represents one day.

```python
temperatures = np.array([18, 21, 25, 17, 16,
                         23, 28, 30, 22, 19,
                         15, 24, 27, 20])

temperatures
```

### Filtering and Counting Days

A condition such as `temperatures > 25` checks each value. Placing this condition inside square brackets selects the matching temperatures.

Using `.sum()` on the condition counts the matching days: True counts as 1 and False as 0.

```python
#Temperatures on days above 25 degrees
temperatures[temperatures > 25]
```

```python
#Number of days above 25 degrees
(temperatures > 25).sum()
```

```python
#Temperatures on days below 20 degrees
temperatures[temperatures < 20]
```

```python
#Number of days below 20 degrees
(temperatures < 20).sum()
```

```python
#Average temperature
avg_temp = temperatures.mean()

avg_temp
```

```python
#Hottest temperature
temperatures.max()
```

```python
#Coldest temperature
temperatures.min()
```

### Hottest and Coldest Days

`.argmax()` finds the position of the highest temperature. `.argmin()` finds the position of the lowest. Adding 1 gives the day number because array positions start at 0. If values are equal, the first matching position is returned.

```python
#Hottest day number - position of the highest value, plus 1
temperatures.argmax() + 1
```

```python
#Coldest day number
temperatures.argmin() + 1
```

```python
#Number of days above the average temperature
(temperatures > avg_temp).sum()
```

## 3. Retail Sales

Each row represents a store. Each column represents a week. There are 3 stores and 4 weeks.

```python
sales = np.array([
    [120, 150, 180, 200],
    [90, 110, 140, 160],
    [200, 220, 250, 300]
])

sales
```

### Totals and Averages

- `.sum(1)` adds across each row to give a total for each store.
- `.sum(0)` adds down each column to give a total for each week.
- `.sum()` adds all values to give overall sales.

`.mean(1)` calculates the average weekly sales for each store.

```python
#Total sales per store - sum of each row
store_totals = sales.sum(1)

store_totals
```

```python
#Total sales per week - sum of each column
week_totals = sales.sum(0)

week_totals
```

```python
#Overall sales
sales.sum()
```

### Best Performing Store and Week

`.argmax()` finds the position of the largest total. Adding 1 gives the store or week number. `.max()` gives the sales value rather than its position.

```python
#Best performing store based on total sales
store_totals.argmax() + 1
```

```python
#Best performing week across all stores
week_totals.argmax() + 1
```

```python
#Average weekly sales per store - mean of each row
sales.mean(1)
```

## 5. Reshape and Transpose

### Reshape

`.reshape()` changes the dimensions of an array. The number of values must stay the same. Here, 6 values are arranged into 2 rows and 3 columns. By default, values are placed row by row.

```python
#Create a simple array
numbers = np.array([1, 2, 3, 4, 5, 6])

numbers
```

```python
#Reshape into 2 rows and 3 columns
reshaped = numbers.reshape(2, 3)

reshaped
```

```python
#Check the new shape
reshaped.shape
```

### Transpose

For a 2D array, `.transpose()` swaps rows and columns. The example changes from 2 rows and 3 columns to 3 rows and 2 columns. Each original row becomes a column.

```python
#Swap the rows and columns
transposed = reshaped.transpose()

transposed
```

```python
#Check the transposed shape
transposed.shape
```

### Comparing the Results

Reshaping to `(3, 2)` places the values row by row. Transposing swaps the rows and columns, so the values appear in a different arrangement.

These examples save the results separately and keep the original array.

```python
#Compare this result with transposed above
reshaped.reshape(3, 2)
```
