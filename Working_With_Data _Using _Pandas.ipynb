.groupby() groups rows with matching values in a column.

.iterrows() loops through a DataFrame one row at a time.

### Ranking Petal Length

.rank(ascending=True) ranks petal lengths from smallest to largest.
Equal values receive the average of their rank positions.

```python
#Add a rank column for petal length

iris["petal_rank"] = iris["petal_length"].rank(ascending = True)

iris.head()
```

### Creating a DataFrame for Each Species

Filter the species column to create separate DataFrames for
setosa, versicolor and virginica.

```python
#Find the species names in the dataset
iris["species"].unique()
```

```python
#Create a separate DataFrame for each species - .copy() lets us work with each new DataFrame independently.
setosa = iris[iris["species"] == "setosa"].copy()

versicolor = iris[iris["species"] == "versicolor"].copy()

virginica = iris[iris["species"] == "virginica"].copy()
```

```python
setosa.head()
```

```python
versicolor.head()
```

```python
virginica.head()
```

### Separating Sepal and Petal Data


Create 2 final df, one for sepal info and one for petal info (with data from all three species in each)

We will select the relevant columns for each DataFrame & keep the species column so each flower can still be identified.

```python
#Create a DataFrame containing sepal measurements

sepal_df = iris[["sepal_length", "sepal_width", "species"]].copy()

sepal_df
```

```python
#Create a DataFrame containing petal measurements 

#Double brackets [[...]] select multiple columns. Both DataFrames retain every row, so they include all three species.

petal_df = iris[["petal_length", "petal_width", "species"]].copy()

petal_df
```

```python
sepal_df["species"].unique()
```

```python
petal_df["species"].unique()
```

## Merging DataFrames

pd.merge() combines DataFrames using matching columns or indexes.

Here, the original row indexes match the sepal and petal measurements
for each flower.

```python
#Merge the DataFrames using their original row indexes

merged_df = pd.merge(
    sepal_df,
    petal_df,
    left_index=True,
    right_index=True
)

merged_df.head()
```

### Removing the Repeated Column

Both DataFrames contained species, so the merge created species_x
and species_y. Drop species_y and rename species_x to species.

```python
#Drop the repeated species column
merged_df = merged_df.drop(columns=["species_y"])

#Rename the remaining species column
merged_df = merged_df.rename(columns={"species_x": "species"})

merged_df.head()
```

## Concatenation

pd.concat() combines DataFrames along rows or columns.

This combines the three species DataFrames back into one DataFrame.

ignore_index = True creates a new row index starting at 0.

```python
#where n is the total number of rows

#Combine all three species into one DataFrame
iris_combined = pd.concat(
    [setosa, versicolor, virginica],
    ignore_index=True
)

iris_combined.head()
```

```python
#Check that all three species are present:
iris_combined["species"].unique()
```

```python
#Count the rows for each species
iris_combined["species"].value_counts()
```
